| **Band Name**                 | **Frequency Range**     | **Wavelength Range**  |
|--------------------------------|------------------------|-----------------------|
| **High Frequency (HF)**        | 1 MHz – 30 MHz         | 10 m – 100 m         |
| **Very High Frequency (VHF)**  | 30 MHz – 300 MHz       | 1 m – 10 m           |
| **Ultra High Frequency (UHF)** | 300 MHz – 3 GHz        | 10 cm – 1 m          |
| **Super High Frequency (SHF)** | 3 GHz – 30 GHz         | 1 cm – 10 cm         |
| **Extremely High Frequency (EHF)** | 30 GHz – 300 GHz   | 1 mm – 1 cm          |
| **Terahertz (THz) Radiation**  | 300 GHz – 3 THz        | 0.1 mm – 1 mm        |

Örneğin, **5 MHz girdi** için bu fonksiyon **[60, "High Frequency (HF)"]** şeklinde bir liste dönecek.

Frekans aralıklarını **[başlangıç frekansı, bitiş frekansı)** olarak düşünebilirsiniz.


# 1. Fonksiyonu tanımla
Aşağıdaki kod hücresine fonksiyonunuzu tanımlayın ve ardından hücreyi çalıştırın. Bu hücre çalışınca freq.py isimli bir python dosyasına fonksiyonunuzu kaydedecek. Bu dosyayı da reponuza ekleyin. Ödevinizi bu dosya ve not defteri ile birlikte gönderin.

```python
%%writefile freq.py

def f2wb(hz):

    c = 3e8
    wavelength = c / hz

    if 1e6 <= hz < 30e6:
        band = "High Frequency (HF)"

    elif 30e6 <= hz < 300e6:
        band = "Very High Frequency (VHF)"

    elif 300e6 <= hz < 3e9:
        band = "Ultra High Frequency (UHF)"

    elif 3e9 <= hz < 30e9:
        band = "Super High Frequency (SHF)"

    elif 30e9 <= hz < 300e9:
        band = "Extremely High Frequency (EHF)"

    elif 300e9 <= hz < 3e12:
        band = "Terahertz (THz) Radiation"

    else:
        band = "Unknown"

    return [wavelength, band]
```
